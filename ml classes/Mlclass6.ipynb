```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
x1=np.array([2,3,4,5,6,7])
x2=np.array([12,13,14,15,16,17])
y=np.array([17,20,23,26,29,30])
m= len(x1)
w1=0
w2=0
b=0
learning_rate=0.0001
epochs=1000

for i in range(epochs):
    y_pred=w1*x1+w2*x2+b
   
    w1d= -(2/m)*sum(y-y_pred*(x1))
    w2d=  -(2/m)*sum(y-y_pred*(x2))
    bd=  -(2/m)*sum(y-y_pred*(1))
    w1=w1-learning_rate*w1d
    w2=w2-learning_rate*w2d
    b= b-learning_rate
    if i%10==0:
           print(f'mse is  {np.mean(y-y_pred**2)}' )

print(f'w1--------{w1}')
print(f'w2--------{w2}')
print(f'b--------{b}')
```
