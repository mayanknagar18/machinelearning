Question 1: Multiple Linear Regression and Coefficient Interpretation
A real-estate company wants to predict house prices using area, number of bedrooms, house age, and location score

The trained model is:

Price = 12 + 0.045 * Area + 6 * Bedrooms - 1.2 * Age + 8 * Location_Score

```python
import numpy as np
import pandas as pd

data= { 'House': [ 'A','B','C','D'],
'Area':[1200,1800,2400,1500],
'Bedrooms':[2,3,4,3],
'Age':[10,5,8,12],
'Location Score':[3,4,5,2]
}
df=pd.DataFrame(data)
df
```

Tasks:

1.Calculate the predicted price for each house using the given equation.
Price = 12 + 0.045 * Area + 6 * Bedrooms - 1.2 * Age + 8 * Location_Score

```python
df["Predicted_Price_in_lakh"] = (
    12 +
    0.045 * df["Area"] +
    6 * df["Bedrooms"] -
    1.2 * df["Age"] +
    8 * df["Location Score"]
)



print(df)
```

Task:
2.Interpret all four feature coefficients while holding the other features constant.

```python
# Area coefficient
print("If Area increases by 1 sq ft, Price increases by", 0.045, "Lakhs")

# Bedrooms coefficient
print("If Bedrooms increase by 1, Price increases by", 6, "Lakhs")

# Age coefficient
print("If Age increases by 1 year, Price decreases by", 1.2, "Lakhs")

# Location Score coefficient
print("If Location Score increases by 1 point, Price increases by", 8, "Lakhs")
```

Task:
4.Explain why a larger coefficient does not necessarily mean a feature is more important when features use different measurement scales.


answer:
A larger coefficient does not always mean that a feature is more important. This is because different features may use different units and measurement scales.

For example, the coefficient of Bedrooms is 6, while the coefficient of Area is 0.045. At first glance, Bedrooms may seem more important because its coefficient is larger. However, Area is measured in square feet and can vary by hundreds or even thousands of square feet, while the number of bedrooms usually changes only by a small amount.

Therefore, a feature with a smaller coefficient can still have a greater impact on the predicted price if its values vary over a larger range. For this reason, coefficient values should not be compared directly when features are measured on different scales.

Conclusion: The importance of a feature depends not only on its coefficient value but also on the scale and range of the feature.
