Question 1: Diagnose Underfitting and Overfitting


```python
# A telecom company develops Decision Tree classifiers to predict customer churn. The following results are obtained:
import pandas as pd
import numpy as np
import seaborn as sns 
data= {'Model':['A','B','C','D'],
       'max_depth':[1,4,8,15],
       'Training Accuracy':[68,87,98  ,100 ]   ,
       'Validation Accuracy':[66,  84, 76, 69]

}
df=pd.DataFrame(data)
df
```

Task : 1 . Identify which models demonstrate underfitting, good fit, and overfitting, with justification.

```python
def check_fit(row):

    train = row['Training Accuracy']
    val = row['Validation Accuracy']
    gap = train - val

    if train < 70 and val < 70:
        return 'Underfitting'

    elif gap > 15:
        return 'Overfitting'

    else:
        return 'Good Fit'


df['model type'] = df.apply(check_fit, axis=1)
```
